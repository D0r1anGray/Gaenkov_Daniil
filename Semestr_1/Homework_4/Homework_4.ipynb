## Задание №1
Напишите функцию, которая принимает на вход строку и проверяет, является ли она валидным транспортным номером (1 буква, 3 цифры, 2 буквы, 2–3 цифры). Обратите внимание, что не все буквы кириллического алфавита используются в транспортных номерах.

Если номер валиден, то функция должна возвращать отдельно номер и регион.

```python
import re

def validate_car_number(car_id):
    letters_rus = 'АВЕКМНОРСТУХ'
    letters_eng = 'ABEKMHOPCTYX'

    pattern_rus = rf'^[{letters_rus}]\d{{3}}[{letters_rus}]{{2}}\d{{2,3}}$'
    pattern_eng = rf'^[{letters_eng}]\d{{3}}[{letters_eng}]{{2}}\d{{2,3}}$'

    car_id = car_id.upper().replace(" ", "")

    if re.match(pattern_rus, car_id) or re.match(pattern_eng, car_id):
        if len(car_id[-3:]) == 3 and car_id[-3:].isdigit():
            number = car_id[:-3]
            region = car_id[-3:]
        else:
            number = car_id[:-2]
            region = car_id[-2:]
        return f"Результат: Номер {number} валиден. Регион: {region}"
    else:
        return "Номер не валиден."

car_number = input()
print(validate_car_number(car_number))
```
